#### The key idea

Virtual environments are a key part of any Python workflow. If you have not started using them now is a good time to start.
When you install packages without a virtual environment they are added to a your "system" Python's. That is problematic for a few reasons:

1. It is not uncommon to have multiple projects require the same package under a different version. 
2. If you install a new package it might update an existing package with a breaking change.
3. (Extra) It is easier to version control your project's dependencies.

#### Creating a virtual environment
You can create a virtual environment using the built-in `venv` module. These instructions assume you're working in the terminal. However, it's also possible to create a virtual environment using the VS Code interface.

### Windows:
```bash
python -m venv name_of_venv
```

### macOS/Linux:
```bash
python3 -m venv name_of_venv
```


This will create a folder called `name_of_venv/` that contains a clean Python installation. You can chance 'name_of_venv' in the examples above to any other name if you wish. For example, here it might make sense to name it deml-venv. You can also add a `.` in front of the name `.venv` or `.daml-venv` to create a hidden folder. Some developers prefer to keep it hidden so the project folder is cleaner.
2. You can pick any target directory for your virtual environment or even leave it blank. TIP: Create it in the root of your (Python) project.
While you could technically place a virtual environment anywhere you want, as `name_of_venv` is just a path to a specific version of Python which you will later use, VSCode likes it if your virtual environment is just below the top folder. I have noticed that it otherwise has difficulties with finding the correct environment in Jupyter Notebooks. Save yourself the trouble and just always place the virtual environment the top folder of your currently opened project. 

#### Activating a virtual environment
Once created, you need to **activate** the environment. VS Code might also automatically offer to activate it for you, but then it doesn't always show that you're working in the name_of_venv in the terminal, so I prefer to activate the venv myself.


##### Windows (Command Prompt):
```bash
name_of_venv\Scripts\activate
```
Windows might give you an error if you have not yet set the acces policy correct: 


You can solve this by running 
```
Set-ExecutionPolicy -ExecutionPolicy RemoteSigned -Scope CurrentUser
```
Which will allow the current user to freely run locally created scripts (such as our Python virtual activation script)

##### Windows (PowerShell):
```bash
name_of_venv\Scripts\Activate.ps1
```

##### macOS/Linux:
```bash
source name_of_venv/bin/activate
```

After activation, you should see `(name_of_venv)` at the beginning of your terminal prompt, indicating the virtual environment is active.

### Working directory, absolute paths and relative paths

We get it, the whole working directory and paths thing can seem mundane when you're eager to dive into the fun parts of coding. But trust me, understanding these concepts now can save you a lot of frustration later on. So, bear with me for a moment and let's break this down:

1. **Working Directory**:

Think of your computer's file system as a massive library. At any given time, you're standing in one of the aisles (a specific location). This location is your "working directory" in programming terms.
In Python, you can see where you're currently standing using `os.getcwd()`. It gives you the current directory or 'aisle' where your script is 'looking' by default.
Jupyter Notebooks: When you open a Jupyter notebook, the directory it starts in (usually where the notebook file is) becomes its working directory, unless you change it.


````python
import os

os.getcwd()
````

````python
# If you're having trouble reading a file check if it's in your working directory

os.listdir()
````


2. **Absolute Path**:

This is the full address of a file or directory in your system. It's like saying, "Go to the library, enter the North Wing, third aisle, top shelf, fifth book from the left". No matter where you're starting from, this path will always lead to the same place.
Example: /Users/john_doe/Documents/my_project/file.txt (on Unix-based systems) or C:\Users\john_doe\Documents\my_project\file.txt (on Windows).

Another example is the path printed above.

3. **Relative Path**:

This is a path given relative to where you're currently standing (your working directory). It's like saying, "From where you are, go two aisles over and check the bottom shelf". The destination changes based on your starting point.
Example: If your working directory is /Users/john_doe/Documents/, the relative path my_project/file.txt refers to the absolute path /Users/john_doe/Documents/my_project/file.txt.
This is super handy when you want your code to be flexible and work in different environments without hardcoding the full path.

````python
with open("data/lab_1_dataset.csv") as f:
    print(f.readline()) # We read the first line of the CSV, which are the column names.
````


**Why does this matter?** When you're reading files, writing data, or importing modules in Python (or any language, really), you need to tell the program where to find things. If you don't understand your working directory or how to specify paths, you'll get errors or, even worse, accidentally modify the wrong files.

So, even though it might seem boring, spending a little time to understand these concepts is a solid investment.

## Intermezzo: Accessing Python and becoming a pro at virtual environment management
### Python in the terminal 
If you access Python in the terminal, you will work and install packages in the activated environment. If there is no activated environment, this is the default Python version. If you have activated a virtual environment yourself, they will be installed in the virtual environment. Easy, right?

### Python (or general shell access) from inside Jupyter notebooks

As you may or may not know, there are also ways to access the shell from inside Jupyter notebooks. However, this is where you should be a bit more careful, especially when mixing this with the use of virtual environments. Basically, you want to be sure you are accessing the right version of Python. 
The difference between % and ! comes down to session context and environment routing. **As a short answer: You should almost always use % (Magic Commands)**.



#### The difference between ! and %

| Feature | `!` (Shell Execution) | `%` (Line Magic) |
| --- | --- | --- |
| **Execution Context** | Spawns a temporary subshell process, runs the command, and closes. | Runs within the active IPython kernel process. |
| **State Persistence** | Directory changes (`!cd folder`) are lost immediately after the cell runs. | Directory changes (`%cd folder`) persist across subsequent notebook cells. |
| **Environment Targeting** | Targets the system PATH default Python/pip, which often points outside the virtual environment. | Explicitly targets the Python executable running the current notebook kernel. |


When you run `!pip install package` inside a VS Code notebook:

1. The `!` operator calls the system's default shell.
2. The shell often routes `pip` to the **global Python installation** or the base system environment, ignoring the active virtual environment selected in the top-right corner of VS Code.
3. The installation succeeds, but running `import package` in the next cell fails with a `ModuleNotFoundError` because the active kernel's environment was never updated.

Using `%pip install package` or `%conda install package`:

* Forces IPython to query `sys.executable` (the exact virtual environment powering the current kernel).
* Installs the package directly into that active virtual environment.
* Instantly makes the newly installed package available to the notebook without needing to restart the kernel.

### Summary 
* **Use `%pip install <package>`** to add packages to the current notebook session.
* **Use `%cd <path>`** to navigate directory folders.
* **Avoid `!` for package management** to prevent installing libraries into the wrong Python environment.

#### 💡 Avoid ambiguity by adding type hints to your code. 

This is one of the best ways to improve your code's readability.

You also get strongly improved auto-complete. 

````python
def greet2(name: str) -> str:
    return f"Hello, {name} but with types"
````

````python
# The types are hints, they can "lie"

def greet3(name: int) -> int:
    return f"{name} is calling this with an int type hint but it's a string"

greet3("john")
````

````python
# Contrived example

def add_numbers(a, b):
    return a + b


result = add_numbers(10, 20)  # This will work as expected
result2 = add_numbers("Hello", "World")  # This will also work, but may not be intended
# result3 = add_numbers("Hello", 20)  This will raise a TypeError at runtime
````

````python
def add_numbers(a: int, b: int) -> int:
    return a + b

result = add_numbers(10, 20)  # This will work as expected
result2 = add_numbers("Hello", "World")  # Type checker will flag this as incorrect
# result3 = add_numbers("Hello", 20)  Type checker will flag this as incorrect too
````
