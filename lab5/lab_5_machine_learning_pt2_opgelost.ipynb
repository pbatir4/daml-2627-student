#### Recap overfitting and underfitting

Overfitting is when the model doesn't learn general patterns from the data but rather focuses on doing really well on the training data. 

❗ Typically it means that the performance on the test set will be a lot worse than that on the training set.

Underfitting is when the model learns a pattern that doesn't significantly capture the details of the training set.

❗ Typically it means that the performance on the test set will be similarly (bad) on as that on the training set.

> 💡 **Uitleg**
>
> `pd.read_csv("data/lab_5_dataset.csv")` leest het CSV-bestand in als een **DataFrame** (een tabel). Het pad is **relatief**: het vertrekt vanuit de map waarin de notebook staat (`lab5/`).

````python
# Split the data into a training and test set
X = travel_dataset.drop(columns="cost") # by convention the variables are called (capital) X
y = travel_dataset["cost"] # By convention what you want to predict is called (small letter) y
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, random_state=42)
print(f"{len(X_train) = } and {len(X_test) = }")
````

> 💡 **Uitleg**
>
> - `travel_dataset.drop(columns="cost")` → een kopie van de tabel **zonder** de kolom `cost`. Dat zijn onze **features** `X` (de input).

##### ❓ Should we apply our preprocessing step to the training dataset, the test dataset, or both?

**Antwoord:**

**Op beide**, maar met een belangrijke nuance:
- Het model verwacht bij voorspellen exact hetzelfde formaat als waarop het getraind is (dezelfde one-hot-kolommen, dezelfde schaal). Dus ook de testset moet getransformeerd worden.
- Maar de preprocessing mag enkel **leren** (`fit`) van de **trainingsset**. De gemiddelden/standaardafwijkingen van de scaler en de lijst categorieën van de encoder moeten uit de trainingsdata komen, en daarna gebruik je die om de testset te **transformeren** (`transform`).

Als je ook op de testset fit, "lekt" informatie uit de testset in je workflow (*data leakage*) en is je test geen eerlijke simulatie meer van nieuwe, ongeziene data. De functie hieronder doet precies die fout (ze fit telkens opnieuw) — dat wordt verderop in de notebook ook besproken.

````python
#Preprocess the data using the following function:
def preprocess_data(data: np.ndarray, categorical_columns: list[str], numeric_columns: list[str]) -> np.ndarray:
    ohe = OneHotEncoder(sparse_output=False) 
    ohe.fit(data[categorical_columns])
    cat_cols_transformed  = ohe.transform(data[categorical_columns])

    scaler = StandardScaler()
    num_cols_transformed = scaler.fit_transform(data[numeric_columns])

    return np.hstack((cat_cols_transformed, num_cols_transformed))  
````

> 💡 **Uitleg**
>
> Deze functie kregen we van de lectoren:
> - `ohe.fit(...)` leert welke categorieën er zijn, `ohe.transform(...)` zet ze om naar 0/1-kolommen. `sparse_output=False` geeft een gewone (dichte) numpy-array terug.
> - `scaler.fit_transform(...)` doet `fit` en `transform` in één keer: van elke kolom wordt het gemiddelde afgetrokken en gedeeld door de standaardafwijking.
> - `np.hstack((a, b))` plakt twee arrays **naast elkaar** (horizontaal) tot één grote array.
> - De `: list[str]` en `-> np.ndarray` zijn *type hints*: ze documenteren welk type verwacht wordt, maar Python controleert dat niet.

#### Welke kolommen gebruiken we als features?

Voor we kunnen preprocessen, moeten we beslissen welke kolommen categorisch en welke numeriek zijn, en welke we **niet** gebruiken:

- `sales_id` → gewoon een volgnummer, zegt niets over de prijs. Een model zou er enkel ruis uit leren.
- `margin` → is *kost voor de reiziger − wat het agentschap betaalt*. Die is dus **berekend uit `cost`** zelf. Dat is *data leakage*: het antwoord zit verstopt in de input (correlatie met `cost` ≈ 0.93).
- `additional_services_cost` en `rating` → worden pas **tijdens/na** de reis bekend. Op het moment dat Ada's ouders een prijs willen berekenen, heb je die nog niet.

De rest kennen we wél op het moment van boeken.


````python
cat_columns = ["country", "membership_status", "destination", "package_Type", "earlybird_discount"]
numeric_columns = ["age", "previous_purchases", "stay_length", "guests", "travel_month", "months_before_travel"]
````

> 💡 **Uitleg**

````python
#Fit a linear regression model and predict both the training and test set
X_train_preprocessed = preprocess_data(X_train, cat_columns, numeric_columns)
X_test_preprocessed = preprocess_data(X_test, cat_columns, numeric_columns)

lin_reg = LinearRegression()
lin_reg.fit(X_train_preprocessed, y_train)

predictions_lin_reg_train = lin_reg.predict(X_train_preprocessed)
predictions_lin_reg_test = lin_reg.predict(X_test_preprocessed)
````

> 💡 **Uitleg**
>
> Dit is het standaardpatroon in scikit-learn, en het komt bij **elk** model terug:
> 1. **Preprocessen**: train- en testset omzetten naar enkel getallen met de functie van hierboven. (⚠️ Zoals gezegd fit deze functie de testset opnieuw — dat lossen we verderop op met een pipeline.)
> 2. **Model aanmaken**: `LinearRegression()` maakt een leeg, nog niet getraind model.
> 3. **Trainen**: `.fit(X, y)` — het model zoekt de beste gewichten zodat de voorspellingen zo dicht mogelijk bij `y_train` liggen.
> 4. **Voorspellen**: `.predict(X)` geeft een numpy-array met één voorspelde kost per rij.
>
> We voorspellen op **beide** sets: de trainingsset om te zien hoe goed het model de data kent die het al zag, en de testset om te zien hoe goed het **generaliseert** naar nieuwe data.

##### ❓ Plot both the predictions of the training and the test set compared to the actual labels. 
You'll see that you basically have to create the same plot multiple times.
Tip: create a function that can help you avoid repetition.

````python
#Your function here
def plot_predictions(y_true, y_pred, title: str):
    """Scatterplot van echte waarden (x) tegenover voorspelde waarden (y)."""
    fig = px.scatter(
        x=y_true,
        y=y_pred,
        labels={"x": "original values", "y": "predicted values"},
        title=title,
        opacity=0.3,
    )
    max_value = max(y_true.max(), y_pred.max())
    fig.add_shape(type="line", x0=0, y0=0, x1=max_value, y1=max_value,
                  line=dict(color="red", dash="dash"))
    return fig
````

> 💡 **Uitleg**
>
> In plaats van de plotcode telkens te kopiëren, steken we ze in een **functie** (het *DRY*-principe: *Don't Repeat Yourself*).
> - `def plot_predictions(y_true, y_pred, title: str):` → een functie met 3 parameters.
> - `px.scatter(x=..., y=...)` → elk punt is één reis: x = echte kost, y = voorspelde kost. `labels` geeft de assen een naam, `opacity=0.3` maakt de punten doorschijnend zodat je ziet waar ze dicht op elkaar liggen.
> - `fig.add_shape(type="line", ...)` → tekent de rode stippellijn **y = x**. Een perfect model zou alle punten exact op die lijn leggen. Hoe verder een punt van de lijn, hoe groter de fout.
> - `return fig` → we geven de figuur terug. Omdat het de laatste expressie in een notebookcel is, toont Jupyter ze automatisch.

````python
#Code to plot the comparison on the training set
plot_predictions(y_train, predictions_lin_reg_train, "Linear regression: predicted vs. original (training set)")
````

````python
#Code to plot the comparison on the test set
plot_predictions(y_test, predictions_lin_reg_test, "Linear regression: predicted vs. original (test set)")
````

#### Using pipelines to avoid mistakes

In machine learning, the preprocessing step is critical to prepare the data for modeling. However, there's a subtle but crucial methodological error in the function shown below. The issue lies in the lines where `OneHotEncoder` and `StandardScaler` are fitted:

```python
def preprocess_data(data: np.ndarray, categorical_columns: list[str], numeric_columns: list[str]) -> np.ndarray:

    ohe = OneHotEncoder(sparse_output=False) 
    ohe.fit(data[categorical_columns]) # The error occurs here!
    cat_cols_transformed  = ohe.transform(data[categorical_columns])


    scaler = StandardScaler()
    num_cols_transformed = scaler.fit_transform(data[numeric_columns]) # And here!

    return np.hstack((cat_cols_transformed, num_cols_transformed))  
```

The key mistake in the preprocessing function provided lies in fitting the OneHotEncoder and StandardScaler to the data inside the function. This could lead to a situation where, when preprocessing the test set, the encoders and scalers are fitted again separately with the test set statistics, which is incorrect. The preprocessing steps that "learn" from the data, such as encoding categorical variables and scaling numerical variables, should be based solely on the training data to avoid introducing bias from the test set into the model.

The proper methodology is to fit the OneHotEncoder and StandardScaler on the training data only. This way, they "learn" the categories of the categorical variables and the distribution (mean and standard deviation) of the numerical variables from the training set. Once fitted, these preprocessors should then be applied to the test data, ensuring that the transformation applied is consistent and does not give the model any information about the test set.

One potential way to solve it is as follows:

```python
def fit_preprocessors(train_data: np.ndarray, categorical_columns: list[str], numeric_columns: list[str]) -> (OneHotEncoder, StandardScaler):
    # Fit the OneHotEncoder and StandardScaler to the training data
    ohe = OneHotEncoder(sparse_output=False)
    scaler = StandardScaler()
    ohe.fit(train_data[categorical_columns])
    scaler.fit(train_data[numeric_columns])
    
    # Return the fitted preprocessors
    return ohe, scaler

def transform_data(data: np.ndarray, categorical_columns: list[str], numeric_columns: list[str], ohe: OneHotEncoder, scaler: StandardScaler) -> np.ndarray:
    # Transform data using the already fitted preprocessors
    cat_cols = ohe.transform(data[categorical_columns])
    num_cols = scaler.transform(data[numeric_columns])
    
    # Return the transformed data
    return np.hstack((cat_cols, num_cols))

```

In scikit-learn, the Pipeline and ColumnTransformer classes offer an idiomatic and streamlined way to chain multiple preprocessing steps and a model into a single workflow. The `Pipeline` class allows you to assemble sequences of transformations and a final model, which simplifies your code and helps prevent common mistakes, such as fitting preprocessing steps to the test data. The `ColumnTransformer` is particularly useful for applying different preprocessing to different columns, such as one-hot encoding for categorical variables and scaling for numerical variables. By combining these tools, you not only reduce the need for manual 'glue' code but also safeguard against the leakage of information from the test set into the training process. We highly recommend you always use this in the scope of this course.



The syntax is quite simple:

*  `make_column_transformer` expects multiple tuples. The `transformer` (preprocessing step) is in the first position of the tuple and the columns you apply the transformer (type: `list[str]`) is in the second position.  
*  `make_pipeline` similarly expects all objects in sequence, so typically you add your preprocessing first followed by the model you want to apply. 

````python
from sklearn.compose import make_column_transformer
from sklearn.pipeline import make_pipeline
````

> 💡 **Uitleg**
>
> - `make_column_transformer` → combineert meerdere preprocessing-stappen, elk op **eigen kolommen**.
> - `make_pipeline` → zet stappen achter elkaar: de output van stap 1 is de input van stap 2, enz.

````python
preprocessing = make_column_transformer(
    (StandardScaler(), numeric_columns),
    (OneHotEncoder(sparse_output=False), cat_columns),
    remainder="drop"
)
````

> 💡 **Uitleg**
>
> - Elke tuple is `(transformer, lijst_met_kolommen)`: de `StandardScaler()` wordt toegepast op `numeric_columns`, de `OneHotEncoder` op `cat_columns`.
> - `remainder="drop"` → alle kolommen die je **niet** vermeldt (hier `sales_id`, `rating`, `margin`, `additional_services_cost`) worden weggegooid. Met `remainder="passthrough"` zouden ze ongewijzigd doorgegeven worden.
> - ⚠️ Deze cel werkt enkel omdat we `numeric_columns` en `cat_columns` hierboven gedefinieerd hebben.

````python
preprocessing.fit_transform(X_train)
````

> 💡 **Uitleg**
>
> Gewoon ter controle: `fit_transform` op de trainingsset geeft een numpy-array met enkel getallen terug — eerst de 6 geschaalde kolommen, dan alle one-hot-kolommen.

````python
lin_reg_pipe = make_pipeline(preprocessing, LinearRegression())
````

> 💡 **Uitleg**
>
> `make_pipeline(preprocessing, LinearRegression())` → één object dat eerst preprocest en daarna het model toepast. Je kan het behandelen alsof het zelf een model is: het heeft gewoon `.fit()` en `.predict()`.

````python
lin_reg_pipe.fit(X_train, y_train)
predictions_lin_reg_train = lin_reg_pipe.predict(X_train)
predictions_lin_reg_test = lin_reg_pipe.predict(X_test)
````

> 💡 **Uitleg**
>
> - `lin_reg_pipe.fit(X_train, y_train)` → fit **eerst** de scaler en encoder op `X_train`, transformeert `X_train`, en traint dan de lineaire regressie. Alles in één keer, en **enkel op trainingsdata**.
> - `lin_reg_pipe.predict(X_test)` → gebruikt de **al geleerde** scaler/encoder (enkel `transform`, geen `fit`!) en voorspelt daarna. Daardoor is de fout van daarnet (opnieuw fitten op de testset) automatisch opgelost.
> - We geven de **ruwe** `X_train`/`X_test` mee (met alle kolommen); de pipeline kiest zelf de juiste kolommen.

That's all!

Convince yourself for a second that is a lot simpler than what we were doing previously, we simply use `make_column_transformer` to indicate what preprocessing we want to apply to which column. Afterwards we put our preprocessing in a pipeline with the model we want to use.

❗ Once you call `model.fit(X_train, y_train)` it both fits the preprocessing and the model in one go. 

Pipelines make it easy to make many different models in one go.

````python
#YOUR FUNCTION HERE
from sklearn.ensemble import RandomForestRegressor
from sklearn.neighbors import KNeighborsRegressor

def train_models(models: dict, preprocessing, X_train, y_train, X_test) -> list[tuple]:
    """Traint elk model in een pipeline en geeft (naam, train-voorspellingen, test-voorspellingen) terug."""
    results = []
    for name, model in models.items():
        pipe = make_pipeline(preprocessing, model)
        pipe.fit(X_train, y_train)
        predictions_train = pipe.predict(X_train)
        predictions_test = pipe.predict(X_test)
        results.append((name, predictions_train, predictions_test))
    return results

models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree": DecisionTreeRegressor(random_state=42),
    "Random Forest": RandomForestRegressor(random_state=42, n_jobs=-1),
    "K Nearest Neighbors": KNeighborsRegressor(),
}

results = train_models(models, preprocessing, X_train, y_train, X_test)
[name for name, _, _ in results]
````

> 💡 **Uitleg**
>
> Stap voor stap:
> - **Imports**: `RandomForestRegressor` (een "bos" van veel beslissingsbomen, elk op een willekeurig deel van de data, waarvan het gemiddelde genomen wordt) en `KNeighborsRegressor` (voorspelt het gemiddelde van de *k* = 5 meest gelijkaardige reizen uit de trainingsset).
> - **`models = {...}`** → een *dictionary*: sleutel = naam (string), waarde = een (nog niet getraind) model.
> - **`for name, model in models.items():`** → `.items()` geeft telkens een paar (sleutel, waarde) terug, dat we meteen uitpakken in `name` en `model`.
> - In de lus maken we per model een **nieuwe pipeline** met dezelfde preprocessing, trainen we ze en voorspellen we train en test.
> - **`results.append((name, ..., ...))`** → let op de dubbele haakjes: de binnenste maken een **tuple**, de buitenste horen bij `append`. Zo krijgen we een lijst van tuples, zoals gevraagd.
> - `n_jobs=-1` → laat de random forest al je CPU-kernen gebruiken (sneller).
> - De laatste regel is een *list comprehension*: `_` is een conventie voor "waarde die ik niet nodig heb".
>

#### Model evaluation

##### ❓ Why is the mean of residuals misleading?


**Antwoord:**

Omdat **positieve en negatieve fouten elkaar opheffen**. Als het model bij de ene reis €500 te hoog zit en bij de andere €500 te laag, is het gemiddelde €0 — alsof het model perfect is, terwijl het er telkens €500 naast zit.

Je ziet dat hier: het gemiddelde van de fouten is bijna 0, maar het histogram toont fouten van honderden euro's in beide richtingen. Het gemiddelde zegt dus enkel of het model **systematisch** te hoog of te laag zit (*bias*), niet **hoe groot** de fouten zijn.

##### ❓ How can we better quantify errors?

**Antwoord:**

Door ervoor te zorgen dat fouten altijd **positief** zijn, zodat ze niet meer tegen elkaar wegvallen:
- de **absolute waarde** nemen → *Mean Absolute Error* (MAE);
- de fout **kwadrateren** → *Mean Squared Error* (MSE), en eventueel daarvan de wortel → *Root Mean Squared Error* (RMSE).

##### Attempt 1: taking the mean of the absolute error (MAE)

````python
np.mean(np.abs(error))
````

> 💡 **Uitleg**
>
> `np.abs(error)` maakt elke fout positief, `np.mean(...)` neemt het gemiddelde. Dit getal lees je als: *"gemiddeld zit het model er zoveel euro naast"*.

````python
px.histogram(np.abs(error), title="distribution of the absolute errors of linear regression on the test set.")
````

This approach is more informative as it provides a clearer picture of how much error is present on average in our predictions.

By considering the MAE, we obtain a useful summary statistic for model performance that is easy to understand and directly interpretable in terms of the problem at hand.

In practice squaring the errors is more common.

##### Attempt 2: Taking the mean of the squared errors (MSE)

Squaring the errors, as shown in the equation: $(-2)² = 4 = (2)²$, ensures that all error values are positive. This method has two main benefits:

This method has two main advantages:

1. Large errors are amplified more than smaller ones, which can be particularly important in cases where larger deviations are less tolerable.
2. MSE is a differentiable function, which makes it mathematically convenient for optimization algorithms used in model training. There are other statistical properties that the MSE can leverage.

Note: point 2 is only for your information. You don't need to know this at all. 


The MSE can be calculated as follows:

````python
np.mean(np.square(error))
````

Since MSE can result in large numbers that are difficult to interpret, we often take the square root to obtain the Root Mean Squared Error (RMSE), which has the same units as the original values:

*Note: the root is $\sqrt(x)$.*

````python
rmse = np.sqrt(np.mean(np.square(error)))
````

````python
rmse
````

> 💡 **Uitleg**
>
> Van binnen naar buiten lezen: `np.square(error)` kwadrateert elke fout → `np.mean` neemt het gemiddelde (= MSE) → `np.sqrt` neemt de wortel (= RMSE). Door de wortel zit je weer in **euro's** in plaats van "euro²". De RMSE is altijd ≥ de MAE; hoe groter het verschil, hoe meer grote uitschieters er in de fouten zitten.

````python
px.histogram(np.square(error), title="distribution of the square errors of linear regression on the test set.")
````

##### ❓ What are downsides of the MSE? There are two, but one is harder to come up with.

**Antwoord:**

1. **Gevoelig voor uitschieters (outliers)**: door het kwadrateren telt één fout van €2000 even zwaar als 400 fouten van €100. Een paar extreme fouten kunnen de MSE dus domineren en een verder goed model slecht doen lijken.
2. **Moeilijk te interpreteren** (de moeilijkere): de MSE is uitgedrukt in **euro²**, een eenheid die geen betekenis heeft. "MSE = 126 000" zegt je intuïtief niets. Daarom nemen we de wortel (RMSE). Bovendien hangt de grootte af van de schaal van het target, dus je kan MSE's van verschillende datasets niet zomaar vergelijken.

##### Summary

Our advice:

* Compute both the MAE and the RMSE. 
* The one you should focus on the most is typically the RMSE. 
* If there are error outliers then the difference between the MAE and the RMSE is likely going to be large. In that case it is typically more interesting to look at the MAE. 
 
Typically, RMSE is preferred because it is more sensitive to large errors, which can be critical in applications where such errors are especially problematic, like in autonomous vehicle guidance systems. However, if your model is prone to outliers, or if large errors are less impactful, MAE can be a more relevant metric. Always consider the specific context of your application when choosing your primary evaluation metric.

##### ❓ Use the pipeline approach discussed above and the mean_absolute_error and mean_squared_error for the following models: RandomForestRegressor, KNeighborsRegressor, DecisionTreeRegressor and LinearRegression. You can build on the function that you created above that returned results for each of the regressors.

````python
#YOUR CODE HERE
def evaluate_models(results: list[tuple], y_train, y_test) -> pd.DataFrame:
    """Berekent MAE en RMSE op train- en testset voor elk model in results."""
    rows = []
    for name, predictions_train, predictions_test in results:
        rows.append({
            "model": name,
            "MAE train": mean_absolute_error(y_train, predictions_train),
            "MAE test": mean_absolute_error(y_test, predictions_test),
            "RMSE train": np.sqrt(mean_squared_error(y_train, predictions_train)),
            "RMSE test": np.sqrt(mean_squared_error(y_test, predictions_test)),
        })
    return pd.DataFrame(rows).set_index("model")

evaluation = evaluate_models(results, y_train, y_test)
evaluation
````

> 💡 **Uitleg**
>
> - We hergebruiken `results` van hierboven, dus we hoeven niets opnieuw te trainen.
> - `for name, predictions_train, predictions_test in results:` → elke tuple heeft 3 elementen, die we meteen uitpakken in 3 variabelen.
> - Per model maken we een **dictionary** (één rij in de tabel). De sleutels worden de kolomnamen.
> - `np.sqrt(mean_squared_error(...))` → MSE omzetten naar RMSE. (Nieuwere scikit-learn-versies hebben ook `root_mean_squared_error`, maar zo zie je wat er gebeurt.)
> - `pd.DataFrame(rows)` → een lijst van dictionaries wordt automatisch een tabel. `.set_index("model")` maakt van de modelnaam de rij-index, zodat de tabel leesbaarder is.

````python
px.bar(evaluation[["RMSE train", "RMSE test"]], barmode="group",
       title="RMSE per model on the training and test set", labels={"value": "RMSE", "variable": "set"})
````

> 💡 **Uitleg**
>
> `evaluation[["RMSE train", "RMSE test"]]` → dubbele vierkante haken selecteren **meerdere kolommen**. `px.bar` met een DataFrame tekent per rij (model) een balk per kolom; `barmode="group"` zet de balken naast elkaar in plaats van op elkaar. Zo zie je in één oogopslag het verschil tussen train en test.
