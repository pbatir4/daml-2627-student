#### Intermezzo: Series and Dataframes are mutable

The first important thing to notice about Pandas dataframe is that they are a combination of Pandas Series (columns).

<center><img src="images\df_vs_series.png"  style="background-color:white;"></center>

**❗ Pandas Series and Dataframes are mutable. This can be a big source of frustration and unexpected behavior. This even differs across pandas versions, so we are going to focus on the newest pandas version.**


```python
# The easiest way to make a dataframe is start from a list of dictionaries

students = [{"name": "Omar", "age": 21}, {"name": "Alisha", "age":19}, {"name": "Joost", "age":18}]
students_df = pd.DataFrame(students)
students_df
```

```python
type(students_df)
```

```python
students_df["name"]
```

```python
type(students_df["name"])
```

```python
student_names = students_df["name"]
student_names
```

```python
# In the newest pandas version, this modifies the series but doesn't modify the dataframe.
student_names[1] = "Katrien"
student_names
```

```python
#We can see our dataframe has remained unchanged.
students_df
```

Previous pandas versions would give the following warning: *"A value is trying to be set on a copy of a slice from a DataFrame"* and would actually update the value in the dataframe. As this is confusing for beginners, it's nice that the newest pandas version removed this. If you are still using an older version, be sure to update.

#### Data exploration: univariate

After that short detour we now know that changing our subset `travel_dataset_subset` means we also change our original dataset. We will continue the exploration of our country, stay_length, age and cost variables.

1. We will start with a univariate analysis, which means we will explore one (uni) variable (variate) at a time. 
2. Later on we will move to two (bi) variables (variate) analyses 
3. We round it up with methods that are able to do multivariate analysis. 

Recall that 

1. **Categorical variables** represent categories or labels (e.g., colors, genders). 
2. **Numeric variables** represent quantities and can be ordered or measured (e.g., age, height). 
3. There is a special case called **Ordinal variables**, these are categories where there is a meaningful order (e.g., clothes sizes: small, medium and large).

Different data types require different techniques:

Numerical + numerical

* Correlation

##### ❓ Carry out these analyses

💻📊💡 TIP: Things like  `dataframe[["column1", "column2"]].groupby("column1").mean()` are valid Pandas. In SQL this would be `select column1, mean(column2) from dataframe group by column 1` 

```python
travel_dataset[["age", "cost", "stay_length"]].corr()
```

There seems to be a strong positive correlation between cost and stay length, meaning that when stay length increases, so does cost, generally speaking.
There seems to be no linear relation between age and cost or age and stay length.

```python
travel_dataset[["cost", "destination", "age", "stay_length"]].groupby("destination").mean()
```

We can see that rips to Sydney are generally cheapest, while Tokyo is most expensive. There does not seem to be a large difference between mean age and stay length for any of the destinations

```python
pd.crosstab(travel_dataset['country'], travel_dataset['destination'])
```

We can now see the number of destinations per country. Most of the journeys are between Canada to Rio. The least popular journey is from Germany to Rome. Overall, the differences between the different countries and destinations is small.
